```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("--- Intermediate Matplotlib & Seaborn Exercise ---")
print("Complete each task by writing the requested plotting code.")
print("--------------------------------------------------")

# Load dataset
tips = sns.load_dataset("tips")

# =====================================================
# Task 1.1 Histogram: Distribution of total_bill
# =====================================================

plt.figure(figsize=(8,5))

plt.hist(
    tips['total_bill'],
    bins=20,
    edgecolor='black'
)

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()

# Insight:
# Most bills fall between 10–25.
# Distribution is right-skewed.


# =====================================================
# Task 1.2 Pie Chart: Smoker Status
# =====================================================

smoker_counts = tips['smoker'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    smoker_counts,
    labels=smoker_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Smoker vs Non-Smoker")

plt.show()

# Insight:
# More non-smokers than smokers.


# =====================================================
# Task 2.1 Scatter Plot
# total_bill vs tip
# =====================================================

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker'
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()

# Insight:
# Positive relationship between bill and tip.


# =====================================================
# Task 2.2 Bar Plot
# Average total bill by day
# =====================================================

plt.figure(figsize=(8,5))

sns.barplot(
    data=tips,
    x='day',
    y='total_bill'
)

plt.title("Average Total Bill by Day")

plt.xlabel("Day")
plt.ylabel("Average Total Bill")

plt.show()

# Insight:
# Saturday generally highest.


# =====================================================
# Task 2.3 Horizontal Bar Plot
# Average tip by time
# =====================================================

avg_tip = (
    tips
    .groupby('time')['tip']
    .mean()
)

plt.figure(figsize=(7,5))

plt.barh(
    avg_tip.index,
    avg_tip.values
)

plt.title("Average Tip by Time")

plt.xlabel("Average Tip")
plt.ylabel("Meal Time")

plt.show()

# Insight:
# Dinner slightly higher.


# =====================================================
# Task 3.1 Multiple Subplots
# =====================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14,10)
)

# --------------------------------

sns.histplot(
    tips['tip'],
    bins=20,
    ax=axes[0,0]
)

axes[0,0].set_title(
    "Histogram of Tips"
)

# --------------------------------

sns.boxplot(
    data=tips,
    x='sex',
    y='total_bill',
    ax=axes[0,1]
)

axes[0,1].set_title(
    "Total Bill by Gender"
)

# --------------------------------

sns.countplot(
    data=tips,
    x='day',
    ax=axes[1,0]
)

axes[1,0].set_title(
    "Customer Count by Day"
)

# --------------------------------

sns.swarmplot(
    data=tips,
    x='smoker',
    y='tip',
    ax=axes[1,1]
)

axes[1,1].set_title(
    "Tip by Smoker Status"
)

plt.tight_layout()

plt.show()

# =====================================================

print("\n--- Exercise Complete! ---")

print(
    "You practiced histogram, pie chart, scatter,"
)

print(
    "bar plot, horizontal bar plot,"
)

print(
    "subplots and EDA interpretation."
)
```
