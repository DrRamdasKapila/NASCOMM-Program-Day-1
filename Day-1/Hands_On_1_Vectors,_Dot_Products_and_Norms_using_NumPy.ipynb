**Creating Vectors & Basic Operations**

```python
import numpy as np

u = np.array([2, 3])
v = np.array([4, 1])

print("u + v =", u + v)    # [6, 4]
print("3 * u =", 3 * u)    # [6, 9]
```

**2. Dot Product**

```python
import numpy as np

u = np.array([2, 3])
v = np.array([4, 1])

dot1 = np.dot(u, v)        # 2*4 + 3*1 = 11
dot2 = u @ v              # Same result

cos_theta = np.dot(u, v) / (
    np.linalg.norm(u) * np.linalg.norm(v)
)

theta = np.arccos(cos_theta)

print("Dot Product (np.dot):", dot1)
print("Dot Product (@):", dot2)
print(f"Angle: {np.degrees(theta):.1f}°")
```

**3. Norms & Cosine Similarity**

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

w = np.array([3, -4])

print("L1:", np.linalg.norm(w, 1))      # 7.0
print("L2:", np.linalg.norm(w, 2))      # 5.0
print("Linf:", np.linalg.norm(w, np.inf))  # 4.0

a = np.array([[1, 2, 3]])
b = np.array([[2, 4, 6]])

print(cosine_similarity(a, b))   # [[1.0]]
```

Create three vectors representing student exam scores in [Math, Science, English].
**bold text**
Compute L2 norms (overall performance) and cosine similarity between pairs.
**bold text**
**Interpret the results.**

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Student Scores [Math, Science, English]

s1 = np.array([85, 90, 88])
s2 = np.array([80, 92, 85])
s3 = np.array([60, 65, 62])

# ----------------------------
# L2 Norms
# ----------------------------

print("L2 Norms")
print("Student 1:", np.linalg.norm(s1))
print("Student 2:", np.linalg.norm(s2))
print("Student 3:", np.linalg.norm(s3))

# ----------------------------
# Cosine Similarities
# ----------------------------

sim12 = cosine_similarity([s1], [s2])
sim13 = cosine_similarity([s1], [s3])
sim23 = cosine_similarity([s2], [s3])

print("\nCosine Similarities")
print("Student 1 vs Student 2 =", sim12[0][0])
print("Student 1 vs Student 3 =", sim13[0][0])
print("Student 2 vs Student 3 =", sim23[0][0])
```

**Alternative (More Elegant)**

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

students = np.array([
    [85, 90, 88],  # Student 1
    [80, 92, 85],  # Student 2
    [60, 65, 62]   # Student 3
])

# L2 Norms
for i, student in enumerate(students, start=1):
    print(f"Student {i} L2 Norm:",
          np.linalg.norm(student))

# Cosine Similarity Matrix
similarity_matrix = cosine_similarity(students)

print("\nCosine Similarity Matrix")
print(similarity_matrix)
```
