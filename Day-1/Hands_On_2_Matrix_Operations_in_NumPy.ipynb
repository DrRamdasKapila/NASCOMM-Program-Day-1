1. Matrix Creation & Basic Operations

```python
import numpy as np

A = np.array([[2, 1],
              [0, 3]])

B = np.array([[1, 4],
              [2, 5]])

print("A + B =\n", A + B)

print("\nA transpose =\n", A.T)

print("\nA @ B =\n", A @ B)   # Matrix multiplication
```

2. Identity Matrix & Matrix Inverse

```python
import numpy as np

I = np.eye(3)

print("Identity Matrix:\n", I)

A = np.array([[2, 1],
              [1, 3]])

A_inv = np.linalg.inv(A)

print("\nA inverse:\n", A_inv)

print("\nA @ A_inv =\n", A @ A_inv)
```

3. Solving Ax = b

```python
import numpy as np

# Solve:
# 2x + y = 5
# x + 3y = 7

A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 7])

x = np.linalg.solve(A, b)

print("Solution:", x)

# Verification
print("Check:", A @ x)
```

Exercise Solution (3×3 Real-World Scenario)
Product Pricing Problem

Suppose a shop sells:

Product A
Product B
Product C

Customer Orders:

Order	Product A	Product B	Product C	Total Bill
1	2	1	1	₹130
2	1	3	2	₹230
3	3	2	4	₹310

Find the price of each product

```python
import numpy as np

A = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [3, 2, 4]
])

b = np.array([
    130,
    230,
    310
])

prices = np.linalg.solve(A, b)

print("Product Prices:")
print("Product A =", prices[0])
print("Product B =", prices[1])
print("Product C =", prices[2])

# Verification
print("\nVerification:")
print(A @ prices)
```
